### 3.1 Datetime Parsing, Indexing & Resampling

```python
# ═══════════════════════════════════════════════════════════════════
#  Parse datetime column and set as index
# ═══════════════════════════════════════════════════════════════════

df_ts = df.copy()

# Step 1: Parse the datetime column
# Hint: pd.to_datetime() — use format='' if you know it, e.g., '%Y-%m'
df_ts[DATE_COLUMN] = pd.to_datetime(df_ts[DATE_COLUMN], errors='coerce')

# Step 2: Check how many dates failed to parse
failed_parse = df_ts[DATE_COLUMN].isna().sum()
print(f"Unparseable dates: {failed_parse}")

# Step 3: Sort chronologically
df_ts = df_ts.sort_values(DATE_COLUMN).reset_index(drop=True)

# Step 4: Set datetime as index
df_ts = df_ts.set_index(DATE_COLUMN)

# Step 5: Resample / aggregate to your chosen frequency
#  Choose frequency — 'D'=daily, 'W'=weekly, 'MS'=monthly start,
#        'QS'=quarterly, 'AS'=annual
FREQ = 'MS'

#  Choose aggregation function — sum(), mean(), last(), max()
# For counts/totals → sum()  |  For rates/averages → mean()
ts = df_ts[[TARGET_COLUMN]].resample(FREQ).sum()

print(f"\nTime series frequency : {FREQ}")
print(f"Number of periods     : {len(ts)}")
print(f"Start date            : {ts.index.min().date()}")
print(f"End date              : {ts.index.max().date()}")
display(ts.head(10))
```

### 3.2 Missing Value Detection & Treatment

```python
# ── Check for NaN values in the time series ──────────────────────────
missing_count = ts[TARGET_COLUMN].isna().sum()
missing_pct   = missing_count / len(ts) * 100
print(f"Missing values: {missing_count} ({missing_pct:.2f}%)")

# ── Check for implicit gaps in the temporal index ────────────────────
# (periods that are missing entirely, not just NaN)
full_idx      = pd.date_range(start=ts.index.min(), end=ts.index.max(), freq=FREQ)
missing_dates = full_idx.difference(ts.index)
print(f"Implicit gap periods : {len(missing_dates)}")

if len(missing_dates) > 0:
    print("Gap periods:", missing_dates.tolist()[:10], "...")

# ── Reindex to ensure a complete, gapless index ──────────────────────
ts = ts.reindex(full_idx)



# Option A — Time-based interpolation (preserves trend):
#   ts[TARGET_COLUMN] = ts[TARGET_COLUMN].interpolate(method='time')
#
# Option B — Linear interpolation:
#   ts[TARGET_COLUMN] = ts[TARGET_COLUMN].interpolate(method='linear')
#
# Option C — Forward fill (uses last known value):
#   ts[TARGET_COLUMN] = ts[TARGET_COLUMN].ffill()
#
# Option D — Seasonal mean imputation (use same period from prior year):
#   # Manual — implement if appropriate
#
# Option E — Drop rows (only if very few are missing):
#   ts = ts.dropna()

ts[TARGET_COLUMN] = ts[TARGET_COLUMN].copy()   # No explicit missing values, so no-op

print(f"\nMissing values after treatment: {ts[TARGET_COLUMN].isna().sum()}")
print(" Missing value treatment complete.")
```

### 3.7 Stationarity Testing: ADF + KPSS

> **Two complementary tests:**  
> - **ADF** — H₀: Unit root exists (non-stationary). Reject if p < 0.05 → stationary.  
> - **KPSS** — H₀: Series is stationary. Reject if p < 0.05 → non-stationary.

```python
def adf_kpss_test(series, label='Series'):
    """Run ADF and KPSS tests and print a combined interpretation."""
    s = series.dropna()

    # ADF Test
    adf_stat, adf_p, adf_lags, adf_nobs, adf_cv, _ = adfuller(s, autolag='AIC')

    # KPSS Test
    kpss_stat, kpss_p, kpss_lags, kpss_cv = kpss(s, regression='c', nlags='auto')

    print(f"\n{'═'*60}")
    print(f"  STATIONARITY TESTS: {label}")
    print(f"{'═'*60}")
    print(f"  {'Test':<22} {'Statistic':>12} {'p-value':>10} {'Verdict':>20}")
    print(f"  {'─'*58}")

    adf_verdict  = '✅ STATIONARY'   if adf_p  < 0.05 else '❌ NON-STATIONARY'
    kpss_verdict = '✅ STATIONARY'   if kpss_p > 0.05 else '❌ NON-STATIONARY'

    print(f"  {'ADF (H0: unit root)':<22} {adf_stat:>12.4f} {adf_p:>10.4f} {adf_verdict:>20}")
    print(f"  {'KPSS (H0: stationary)':<22} {kpss_stat:>12.4f} {kpss_p:>10.4f} {kpss_verdict:>20}")

    print(f"\n  ADF Critical Values : 1%={adf_cv['1%']:.3f}  5%={adf_cv['5%']:.3f}  10%={adf_cv['10%']:.3f}")
    print(f"  KPSS Critical Values: 1%={kpss_cv['1%']:.3f}  5%={kpss_cv['5%']:.3f}  10%={kpss_cv['10%']:.3f}")

    # Consensus
    adf_stat_   = adf_p  < 0.05
    kpss_stat_  = kpss_p > 0.05
    print(f"\n  ┌─ CONSENSUS ───────────────────────────────────────┐")
    if adf_stat_ and kpss_stat_:
        print(f"  │  ✅ Both tests agree: STATIONARY                  │")
    elif not adf_stat_ and not kpss_stat_:
        print(f"  │  ❌ Both tests agree: NON-STATIONARY               │")
    elif adf_stat_ and not kpss_stat_:
        print(f"  │  ⚠️  Trend-stationary (possibly difference)        │")
    else:
        print(f"  │  ⚠️  Difference-stationary (possible unit root)    │")
    print(f"  └───────────────────────────────────────────────────┘")
    print(f"{'═'*60}")

    return {'adf_p': adf_p, 'kpss_p': kpss_p,
            'adf_stationary': adf_stat_, 'kpss_stationary': kpss_stat_}

print("Stationarity test function defined.")
```

```python
# ── Test 1: Original (or transformed) series ─────────────────────────
r0 = adf_kpss_test(ts_model, label='Original / Transformed Series')

# ── Test 2: First difference ──────────────────────────────────────────
ts_diff1 = ts_model.diff(1)
r1 = adf_kpss_test(ts_diff1.dropna(), label='1st Difference (d=1)')

# ── Test 3: Seasonal difference (lag = PERIOD) ────────────────────────
ts_sdiff = ts_model.diff(PERIOD)
r_s = adf_kpss_test(ts_sdiff.dropna(), label=f'Seasonal Difference (lag={PERIOD})')

# ── Test 4: First + Seasonal difference ──────────────────────────────
ts_diff1s = ts_model.diff(PERIOD).diff(1)
r1s = adf_kpss_test(ts_diff1s.dropna(), label=f'1st + Seasonal Difference')



# Based on test results above, select the stationary series
# and record the differencing parameters d and D

d = 1   # Non-seasonal differences needed (0, 1, or 2)
D = 0   # Seasonal differences needed    (0 or 1)

# Apply the identified differencing
if D > 0:
    ts_stationary = ts_model.diff(PERIOD)
else:
    ts_stationary = ts_model.copy()

if d > 0:
    ts_stationary = ts_stationary.diff(d)

ts_stationary = ts_stationary.dropna()

print(f"\nFinal differencing: d={d}, D={D}")
print(f"Stationary series length: {len(ts_stationary)}")
```

> **Q3.** Why is stationarity a **fundamental requirement** for ARIMA-class models? What theoretical problem arises when fitting ARIMA to a non-stationary series?

> ▶ Stationarity is a fundamental requirement for ARIMA-class models because these models assume that the statistical properties of the time series, such as mean, variance, and autocorrelation, remain constant over time. ### 4.2 Model Selection via AIC / BIC Grid Search

```python
# Define the search space for SARIMA parameters
# Keep ranges small (0–2) to avoid long computation times

p_range = range(0, 3)   # e.g., range(0, 3)
q_range = range(0, 3)
P_range = range(0, 3)
Q_range = range(0, 3)

# d and D fixed from stationarity tests
D_val = D
d_val = d

results_list = []

total = len(p_range) * len(q_range) * len(P_range) * len(Q_range)
print(f"Grid search over {total} combinations...")

for p_val, q_val, P_val, Q_val in itertools.product(p_range, q_range, P_range, Q_range):
    try:
        model = SARIMAX(
            train,
            order=(p_val, d_val, q_val),
            seasonal_order=(P_val, D_val, Q_val, PERIOD),
            enforce_stationarity=False,
            enforce_invertibility=False
        ).fit(disp=False)

        results_list.append({
            'p': p_val, 'd': d_val, 'q': q_val,
            'P': P_val, 'D': D_val, 'Q': Q_val,
            'AIC': round(model.aic, 3),
            'BIC': round(model.bic, 3),
            'HQIC': round(model.hqic, 3)
        })
    except Exception:
        pass

grid_df = pd.DataFrame(results_list).sort_values('AIC').reset_index(drop=True)

print("\n📊 Top 10 Models by AIC:")
print(grid_df.head(10).to_string(index=False))

# ── Identify best model ───────────────────────────────────────────────
best_row = grid_df.iloc[0]
print(f"\n🏆 Best model: SARIMA({best_row.p},{best_row.d},{best_row.q})"
      f"({best_row.P},{best_row.D},{best_row.Q},{PERIOD})")
print(f"   AIC={best_row.AIC}  |  BIC={best_row.BIC}  |  HQIC={best_row.HQIC}")
```

####  Model Selection Justification

> **Q1.** What is the difference between **AIC** and **BIC**? Which penalizes model complexity more heavily, and why does this matter?

> ▶ AIC (Akaike Information Criterion) and BIC (Bayesian Information Criterion) are both measures used for comparing statistical models, aiming to balance goodness-of-fit with model complexity. AIC is calculated as `2k - 2ln(L)`, while BIC is `kln(n) - 2ln(L)`, where `k` is the number of parameters, `L` is the maximum likelihood, and `n` is the number of observations. BIC penalizes model complexity more heavily than AIC, especially for larger datasets, because the term `ln(n)` grows faster than `2` for `n > 7`. This matters because BIC tends to select simpler models, which can be beneficial in preventing overfitting and promoting better generalization to new data.

However, looking at the full `grid_df`, the model `SARIMA(1,1,0)(0,0,0,12)` has an AIC of -306.069 but a lower (more negative) BIC of -302.018, making it the BIC-best model. I would choose the `SARIMA(1,1,0)(0,0,0,12)` model because BIC's heavier penalty on complexity often leads to more parsimonious models that are less likely to overfit and tend to perform better on out-of-sample data, despite a slightly higher AIC.

### 5.3 Evaluation Metrics

```python
# ── Helper: custom metrics ─────────────────────────────────────────────
def smape(actual, forecast):
    """Symmetric Mean Absolute Percentage Error (0–200%)."""
    return 100 * np.mean(2 * np.abs(actual - forecast) /
                         (np.abs(actual) + np.abs(forecast) + 1e-9))

def mape(actual, forecast):
    """Mean Absolute Percentage Error."""
    return 100 * np.mean(np.abs((actual - forecast) / (actual + 1e-9)))

# ── Compute metrics ───────────────────────────────────────────────────
actual_vals   = test_actual.values
forecast_vals = forecast_mean.values

mae_val   = mean_absolute_error(actual_vals, forecast_vals)
rmse_val  = np.sqrt(mean_squared_error(actual_vals, forecast_vals))
mape_val  = mape(actual_vals, forecast_vals)
smape_val = smape(actual_vals, forecast_vals)

# ── Naïve baseline: last observed value repeated ─────────────────────
naive_forecast = np.full(FORECAST_HORIZON, train_plot.iloc[-1])
mae_naive  = mean_absolute_error(actual_vals, naive_forecast)
rmse_naive = np.sqrt(mean_squared_error(actual_vals, naive_forecast))

# ── Seasonal naïve: repeat same period from last cycle ───────────────
seasonal_naive = train_plot.values[-PERIOD:] if len(train_plot) >= PERIOD else naive_forecast
seasonal_naive = np.tile(seasonal_naive, int(np.ceil(FORECAST_HORIZON / PERIOD)))[:FORECAST_HORIZON]
mae_snaive  = mean_absolute_error(actual_vals, seasonal_naive)
rmse_snaive = np.sqrt(mean_squared_error(actual_vals, seasonal_naive))

# ── Print comparison table ────────────────────────────────────────────
metrics_df = pd.DataFrame({
    'Model'  : ['SARIMA (Your Model)', 'Naïve Baseline', 'Seasonal Naïve'],
    'MAE'    : [mae_val, mae_naive, mae_snaive],
    'RMSE'   : [rmse_val, rmse_naive, rmse_snaive],
    'MAPE %' : [mape_val, mape(actual_vals, naive_forecast),
                mape(actual_vals, seasonal_naive)],
    'SMAPE %': [smape_val, smape(actual_vals, naive_forecast),
                smape(actual_vals, seasonal_naive)]
})
metrics_df = metrics_df.round(4)
print("\n" + "=" * 72)
print(" MODEL PERFORMANCE COMPARISON")
print("=" * 72)
print(metrics_df.to_string(index=False))
print("=" * 72)
print("\n📌 Lower values = better performance for all metrics.")
print("   MAPE > 100% or SMAPE > 100% indicates very poor fit.")

# ── Skill score (vs naïve): how much better than random? ─────────────
skill_mae  = 1 - mae_val  / mae_naive
skill_rmse = 1 - rmse_val / rmse_naive
print(f"\n  Skill Score (vs Naïve) — MAE : {skill_mae*100:.1f}% improvement")
print(f"  Skill Score (vs Naïve) — RMSE: {skill_rmse*100:.1f}% improvement")
```

#### Metric Interpretation

> **Q1.** Explain what **MAE** and **RMSE** measure. How do they differ in how they treat large errors?

> ▶ MAE (Mean Absolute Error) tells us the average size of our prediction errors, ignoring the direction. RMSE (Root Mean Squared Error) also measures average error, but it gives more weight to larger errors because it squares them before averaging. This means RMSE is more sensitive to big mistakes than MAE.

> **Q2.** Explain the difference between **MAPE** and **SMAPE**. When would SMAPE be preferred?

>  ▶ MAPE (Mean Absolute Percentage Error) is the average percentage error, making it easy to understand as a percentage of the actual value. However, MAPE can be problematic if actual values are zero or very small. SMAPE (Symmetric Mean Absolute Percentage Error) is similar but uses an average of the actual and forecast in the denominator, making it more robust when actual values are close to zero, and it keeps the percentage interpretation.

```python
# ═══════════════════════════════════════════════════════════════════
# FORMAL STATISTICAL TESTS ON RESIDUALS
# ═══════════════════════════════════════════════════════════════════

print("=" * 60)
print(" FORMAL RESIDUAL TESTS")
print("=" * 60)

# ── Test 1: Ljung-Box (autocorrelation in residuals) ─────────────────
lb_result = acorr_ljungbox(residuals.dropna(),
                           lags=[PERIOD, 2*PERIOD], return_df=True)
print("\n1. Ljung-Box Test (H0: No autocorrelation in residuals)")
print(lb_result)
for _, row in lb_result.iterrows():
    verdict = "✅ White noise" if row['lb_pvalue'] > 0.05 else "⚠️  Autocorrelation present"
    print(f"   Lag {int(row.name):>3}: p={row['lb_pvalue']:.4f} → {verdict}")

# ── Test 2: Shapiro-Wilk (normality of residuals) ────────────────────
# Use first 5000 observations (Shapiro-Wilk limit)
sw_sample = residuals.dropna().values[:5000]
sw_stat, sw_p = shapiro(sw_sample)
sw_verdict = "✅ Normal" if sw_p > 0.05 else "⚠️  Non-normal"
print(f"\n2. Shapiro-Wilk Test (H0: Residuals are normal)")
print(f"   Statistic={sw_stat:.4f}, p-value={sw_p:.4f} → {sw_verdict}")

# ── Test 3: D'Agostino-Pearson (normality, for larger samples) ───────
dp_stat, dp_p = normaltest(residuals.dropna().values)
dp_verdict = "✅ Normal" if dp_p > 0.05 else "⚠️  Non-normal"
print(f"\n3. D'Agostino-Pearson Test (H0: Residuals are normal)")
print(f"   Statistic={dp_stat:.4f}, p-value={dp_p:.4f} → {dp_verdict}")

# ── Residual summary statistics ───────────────────────────────────────
print(f"\n4. Residual Summary")
print(f"   Mean     : {residuals.mean():.6f}  (should be ≈ 0)")
print(f"   Std Dev  : {residuals.std():.4f}")
print(f"   Skewness : {residuals.skew():.4f}  (should be ≈ 0)")
print(f"   Kurtosis : {residuals.kurt():.4f}  (excess; should be ≈ 0)")
```

> *   **Limited Data Length:** While sufficient for demonstrating SARIMA, the dataset consists of only 64 monthly observations. ### 6.3 Overall Conclusion

This time series forecasting project for Elenest Resort’s Total_Income successfully followed the CRISP-DM methodology. In the Business Understanding phase, the main goal was identified as improving financial planning by forecasting the resort’s future income. The project aimed to predict the next 6 months of income while achieving a forecast accuracy with less than 10% MAPE.

During the Data Understanding phase, the dataset was analyzed to better understand the behavior of the resort’s income over time. The results showed a clear upward trend and seasonal patterns, meaning that the income changes depending on certain months or seasons. Because of these patterns, the SARIMA model was identified as the most suitable forecasting method.

In the Data Preparation phase, the dataset was cleaned and transformed before modeling. The date column was converted into a proper datetime format and used as the index of the dataset. No missing values or major outliers were found. STL decomposition confirmed that the dataset had strong trend and seasonal components. To improve the quality of the model, a log1p transformation and differencing were applied to make the data stationary. The ADF and KPSS tests confirmed that one non-seasonal differencing (d=1) was enough, while seasonal differencing (D=0) was not necessary.

In the Modeling phase, several SARIMA parameter combinations were tested using grid search. The SARIMA(2,1,2)(0,0,0,12) model was selected because it achieved the lowest AIC value, meaning it provided the best overall fit among the tested models. Although some coefficients were statistically insignificant, the model still produced strong forecasting performance.

The Evaluation phase showed that the chosen model performed very well. The 6-month forecast successfully kept all actual income values within the 95% confidence interval. The model achieved a MAE of approximately ₱21,007, RMSE of approximately ₱22,998, MAPE of approximately 2.59%, and SMAPE of approximately 2.63%. These results indicate that the forecasts were highly accurate, with only around 2.6% average error, which is well below the target of 10% MAPE. Compared to the Naïve and Seasonal Naïve models, the SARIMA model performed significantly better. Residual analysis also showed no major autocorrelation or heteroscedasticity, meaning the model captured the data patterns effectively. However, the residuals were not perfectly normally distributed, which may slightly affect the reliability of the confidence intervals.

For the Deployment phase, a 12-month forecast from May 2026 to April 2027 was generated using the final model. The forecast suggests that Elenest Resort will maintain a stable and high income, generally ranging from ₱800,000 to ₱830,000. These forecasts can help management improve staffing decisions during peak months and strengthen marketing strategies during slower periods. Since confidence intervals become wider over time, short-term forecasts are more reliable for operational planning, while long-term forecasts are better used for strategic decision-making.

Overall, the project proved that SARIMA is an effective model for forecasting Elenest Resort’s income. Future improvements may include testing other transformations or models to address the non-normality of residuals, simplifying the model by removing insignificant parameters, and incorporating external variables such as holidays, weather conditions, or tourism demand to further improve forecasting accuracy.
